## Section 1: Read the delivery evidence

A retail notification service sends every shipment update through email and SMS. The user story requires routing only through a shopper's selected channels while preserving required-field validation and renderer behavior.

The delivery chain is `story-planner -> human approval -> feature-implementer -> acceptance-reviewer -> final verifier`. Planning and review are read-only. Production ownership is limited to `preferences.py` and `router.py`; tests are immutable.

```python
EVENT = {
    "order_id": "A100",
    "customer_name": "Riley",
    "carrier": "SwiftShip",
    "tracking_url": "https://track.example/A100",
}
CHANNEL_ORDER = ("email", "sms")
REQUIRED_FIELDS = ("order_id", "customer_name", "carrier", "tracking_url")
OWNED_FILES = {"preferences.py", "router.py"}
IMMUTABLE_FILES = {"tests/test_preferences.py", "tests/test_renderers.py"}
print("Story contract loaded: 6 acceptance criteria, 2 owned production files.")
```

## Section 2: Define the specialist agents

Give each agent the least authority needed for its role. The planner inspects evidence, the implementer changes only approved production files, and the reviewer executes checks without repairing its own findings.

```python
AGENT_SPECS = {
    "story-planner": {
        "description": "Analyzes a user story, repository boundaries, and tests before implementation. Use first for feature planning.",
        "tools": {"Read", "Glob", "Grep"},
        "model": "haiku",
        "prompt": "Analyze the story and repository evidence. Return a bounded plan with criteria, allowed files, risks, and exact checks. Do not edit files or claim unexecuted checks passed.",
    },
    "feature-implementer": {
        "description": "Implements an approved bounded Python feature and runs focused checks. Use only after plan approval.",
        "tools": {"Read", "Glob", "Grep", "Edit", "Bash"},
        "model": "sonnet",
        "prompt": "Implement only the approved story within explicit ownership. Preserve tests and unrelated behavior. Run focused checks and report changed files plus real results.",
    },
    "acceptance-reviewer": {
        "description": "Reviews a completed feature against its story, diff boundaries, and executable checks. Use after implementation.",
        "tools": {"Read", "Glob", "Grep", "Bash"},
        "model": "sonnet",
        "prompt": "Independently inspect the story, changed files, implementation, and immutable tests. Run focused and full checks. Return PASS or FAIL per criterion. Do not edit files.",
    },
}
```

```python
EXPECTED_AGENTS = {
    "story-planner": ({"Read", "Glob", "Grep"}, "haiku"),
    "feature-implementer": ({"Read", "Glob", "Grep", "Edit", "Bash"}, "sonnet"),
    "acceptance-reviewer": ({"Read", "Glob", "Grep", "Bash"}, "sonnet"),
}

def contains_todo(value):
    if isinstance(value, str):
        return "todo" in value.lower()
    if isinstance(value, dict):
        return any(contains_todo(item) for item in value.values())
    if isinstance(value, (set, list, tuple)):
        return any(contains_todo(item) for item in value)
    return False

def validate_agents(specs):
    errors = []
    if set(specs) != set(EXPECTED_AGENTS):
        errors.append("Define exactly the three required agents.")
    for name, (tools, model) in EXPECTED_AGENTS.items():
        spec = specs.get(name, {})
        if spec.get("tools") != tools:
            errors.append(f"{name} tools must be {sorted(tools)}.")
        if spec.get("model") != model:
            errors.append(f"{name} must use model {model}.")
        if len(spec.get("description", "").split()) < 8:
            errors.append(f"{name} needs a specific delegation description.")
        if len(spec.get("prompt", "").split()) < 15:
            errors.append(f"{name} needs a specific operating prompt.")
        if contains_todo(spec):
            errors.append(f"{name} still contains TODO text.")
    return errors

agent_errors = validate_agents(AGENT_SPECS)
print("Agent contracts valid." if not agent_errors else "\n".join(agent_errors))
```

## Section 3: Approve a bounded delivery plan

Map implementation ownership and verification before production behavior changes. The approval record is a hard dependency, not a completion claim generated after the work.

```python
DELIVERY_PLAN = {
    "allowed_files": {"preferences.py", "router.py"},
    "immutable_files": {"tests/test_preferences.py", "tests/test_renderers.py"},
    "focused_check": "python3 -m unittest tests.test_preferences",
    "full_check": "python3 -m unittest",
    "acceptance_cases": {"absent", "empty", "invalid", "single-channel", "reordered"},
    "human_approval": "A human approved scope, immutable tests, edge cases, and exact verification commands before implementation.",
    "approved_before_implementation": True,
}
```

```python
def validate_plan(plan):
    errors = []
    if plan.get("allowed_files") != OWNED_FILES:
        errors.append("Plan ownership must be preferences.py and router.py.")
    if plan.get("immutable_files") != IMMUTABLE_FILES:
        errors.append("Both supplied test files must remain immutable.")
    if plan.get("focused_check") != "python3 -m unittest tests.test_preferences":
        errors.append("Plan needs the focused preference check.")
    if plan.get("full_check") != "python3 -m unittest":
        errors.append("Plan needs the full acceptance check.")
    required_cases = {"absent", "empty", "invalid", "single-channel", "reordered"}
    if plan.get("acceptance_cases") != required_cases:
        errors.append("Plan must cover every required preference case.")
    if not plan.get("approved_before_implementation"):
        errors.append("Human approval must be recorded before implementation.")
    if contains_todo(plan.get("human_approval", "")) or len(plan.get("human_approval", "").split()) < 8:
        errors.append("Record an observable human approval decision.")
    return errors

plan_errors = validate_plan(DELIVERY_PLAN)
print("Bounded plan approved." if not plan_errors else "\n".join(plan_errors))
```

## Section 4: Implement within ownership

Implement channel selection and route only to selected renderers. Preserve canonical email-then-SMS order, required-field validation, function signatures, and standard-library-only execution.

```python
def enabled_channels(event):
    requested = event.get("preferred_channels")
    if requested is None:
        return CHANNEL_ORDER
    if not isinstance(requested, list):
        raise ValueError("preferred_channels must be a list")
    unknown = sorted(set(requested) - set(CHANNEL_ORDER))
    if unknown:
        raise ValueError(f"Unsupported preferred channel: {', '.join(unknown)}")
    return tuple(channel for channel in CHANNEL_ORDER if channel in requested)

def render_email(event):
    return {"subject": f"Order {event['order_id']} shipped", "body": f"Hello {event['customer_name']}, track at {event['tracking_url']}"}

def render_sms(event):
    return f"Your order {event['order_id']} shipped with {event['carrier']}. {event['tracking_url']}"

def validate_event(event):
    missing = [field for field in REQUIRED_FIELDS if field not in event]
    if missing:
        raise ValueError(f"Missing shipment fields: {', '.join(missing)}")

def notify(event):
    validate_event(event)
    renderers = {"email": render_email, "sms": render_sms}
    return {channel: renderers[channel](event) for channel in enabled_channels(event)}
```

## Section 5: Execute independent acceptance evidence

The reviewer does not infer success from the implementation. It runs the immutable behavior checks and reports observed results.

```python
class PreferenceRoutingTest(unittest.TestCase):
    def test_defaults_to_both_channels(self):
        self.assertEqual(list(notify(EVENT)), ["email", "sms"])

    def test_routes_only_to_selected_channel(self):
        self.assertEqual(list(notify({**EVENT, "preferred_channels": ["sms"]})), ["sms"])

    def test_uses_canonical_order(self):
        self.assertEqual(list(notify({**EVENT, "preferred_channels": ["sms", "email"]})), ["email", "sms"])

    def test_empty_preferences_disable_notifications(self):
        self.assertEqual(notify({**EVENT, "preferred_channels": []}), {})

    def test_rejects_unsupported_channel(self):
        with self.assertRaisesRegex(ValueError, "Unsupported preferred channel: push"):
            notify({**EVENT, "preferred_channels": ["push"]})

    def test_rejects_non_list_preferences(self):
        with self.assertRaisesRegex(ValueError, "preferred_channels must be a list"):
            notify({**EVENT, "preferred_channels": "sms"})

    def test_preserves_required_field_validation(self):
        incomplete = {key: value for key, value in EVENT.items() if key != "tracking_url"}
        with self.assertRaisesRegex(ValueError, "tracking_url"):
            notify(incomplete)

class RendererTest(unittest.TestCase):
    def test_email_preserves_expected_content(self):
        result = render_email(EVENT)
        self.assertEqual(result["subject"], "Order A100 shipped")
        self.assertIn("Hello Riley", result["body"])
        self.assertIn(EVENT["tracking_url"], result["body"])

    def test_sms_preserves_expected_content(self):
        result = render_sms(EVENT)
        self.assertIn("order A100", result)
        self.assertIn("SwiftShip", result)
        self.assertLessEqual(len(result), 160)

def run_acceptance():
    suite = unittest.TestSuite()
    loader = unittest.defaultTestLoader
    suite.addTests(loader.loadTestsFromTestCase(PreferenceRoutingTest))
    suite.addTests(loader.loadTestsFromTestCase(RendererTest))
    stream = io.StringIO()
    result = unittest.TextTestRunner(stream=stream, verbosity=2).run(suite)
    print(stream.getvalue())
    return result

acceptance_result = run_acceptance()
```

## Section 6: Run the independent gate

The final verifier combines contract structure, plan approval, ownership, and executed tests. Completion requires all evidence at once.

```python
errors = validate_agents(AGENT_SPECS) + validate_plan(DELIVERY_PLAN)
if not acceptance_result.wasSuccessful():
    errors.append("Full acceptance suite failed.")

if errors:
    print("CUSTOM AGENT COLAB LAB INVALID")
    for error in errors:
        print(f"- {error}")
else:
    artifact_root = Path("custom-agent-feature-delivery-artifacts/.claude/agents")
    artifact_root.mkdir(parents=True, exist_ok=True)
    for name, spec in AGENT_SPECS.items():
        tools = ", ".join(sorted(spec["tools"]))
        content = f"---\nname: {name}\ndescription: {spec['description']}\ntools: {tools}\nmodel: {spec['model']}\n---\n\n{spec['prompt']}\n"
        (artifact_root / f"{name}.md").write_text(content, encoding="utf-8")
    print("CUSTOM AGENT COLAB LAB VALID")
    print("Agents: story-planner -> feature-implementer -> acceptance-reviewer")
    print("Owned implementation: preferences.py, router.py")
    print("Human approval: recorded before implementation")
    print(f"Full acceptance suite: {acceptance_result.testsRun} tests passed")
```
